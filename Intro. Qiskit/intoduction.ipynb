```python
import qiskit
print(qiskit.__version__)
```

```python
from qiskit import QuantumCircuit
```

```python
qc = QuantumCircuit(2)
qc.x(0)
qc.draw('mpl')
```

```python
qc = QuantumCircuit(1)
qc.z(0)
qc.draw('mpl')
```

```python
qc = QuantumCircuit(1)
qc.h(0)
qc.draw('mpl')
```

```python
qc = QuantumCircuit(2)
qc.cx(0,1)
qc.draw('mpl')
```

```python
qc = QuantumCircuit(2)
qc.h(0)
qc.cx(0,1)
qc.draw('mpl')
```

```python
# adding classical bits to the circuit
qc = QuantumCircuit(2,2)
qc.h(0)
qc.cx(0,1)
qc.draw('mpl')
```

```python
qc = QuantumCircuit(2,2)
qc.h(0)
qc.cx(0,1)
qc.measure([0,1],[0,1])
qc.draw('mpl')
```

```python
#transpile the circuit for a specific backend
from qiskit_aer import Aer
from qiskit import transpile
sim_backend = Aer.get_backend('qasm_simulator')
compiled = transpile(qc, sim_backend)
```

```python
#Run the circuit on a statevector simulator backend and get the result
job = sim_backend.run(compiled, shots=1000)
result = job.result()
counts = result.get_counts(compiled)
print("Total count for 00 and 11 are:",counts)
```

```python
from qiskit.visualization import plot_histogram
plot_histogram(counts)
```
